```python
import requests
from bs4 import BeautifulSoup

url = "https://www.google.com"

response = requests.get(url)

# Check response
print(response.status_code)

# Parse HTML
soup = BeautifulSoup(response.text, "html.parser")

# Get page title
print("Title:", soup.title.text)

# Get all text
print(soup.get_text())
```
